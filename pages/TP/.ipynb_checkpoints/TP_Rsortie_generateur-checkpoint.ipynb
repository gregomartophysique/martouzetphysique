```python
E0 = 5.00
R = np.array([0,10,20,30,40,50,60,75,100,250,500,1000])
U = np.array([0.005,0.837,1.43,1.875,2.222,2.499,2.726,2.999,3.331,4.163,4.54,4.75])
```

```python
plt.plot( R, U, '+')
plt.grid()
plt.xlabel(r'$R$ ($\Omega$)')
plt.ylabel(r'$U$ ($V$)')
plt.show()
```

```python
Y = R*(E0-U)

coef = np.polyfit( U, Y, deg=1 )

print( coef )

plt.plot( U, Y, '+' )
plt.plot( U, coef[0]*U+coef[1], '--')
plt.xlabel('$U$ (V)')
plt.ylabel('$Y$ (V$\\Omega$)')
plt.grid()
plt.show()
```
